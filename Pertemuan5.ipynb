## Topik

Visualisasi Data










```python
import matplotlib.pyplot as plt

bulan = ["Jan", "Feb", "Mar", "Apr"]
penjualan = [100,120,150,130]

plt.figure(figsize=(6,4))
plt.bar(bulan,penjualan)
plt.title("Grafik Penjualan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah Penjualan")
plt.show()

plt.figure(figsize=(6,4))
plt.plot(bulan, penjualan, marker ='o')
plt.title("Trend Penjualan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah Penjualan")
plt.show()
```
