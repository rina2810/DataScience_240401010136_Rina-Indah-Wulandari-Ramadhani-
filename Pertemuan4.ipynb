## Topik
 Statistika Dasar dan Analisis Data










```python
import pandas as pd

df = pd.DataFrame({
    "Nilai": [80,85,90,75,88]

})

print(df)

print("\nMean :", df["Nilai"].mean())
print ("Median :", df ["Nilai"].median())
print ("Maksimum :", df["Nilai"].max())
print ("Minimum", df["Nilai"].min())
```

## Kesimpulan

Pada pertemuan ini saya mempelajari analisis statistik dasar menggunakan mean, median, nilai maksium dan minum.
