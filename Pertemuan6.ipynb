Persiapan Data










```python
from IPython.core.interactiveshell import default_banner
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.DataFrame({
    "Umur" : [20,21,22,23],
    "Nilai ":[70,80,90,85]
})

print("Data Awal")
print(df)

scaler = MinMaxScaler()

df["Nilai_Normalisasi"] = scaler.fit_transform(df[["Nilai"]])

print("\nData Setelah Normalisasi")
print(df)
```
