## Topik
 Data Cleaning : Missing, Outlier dan Ekstraksi Data

```python
import numpy as np
import pandas as pd

df = pd.DataFrame ({
    "Nama" : ["Andi", "Budi", "Citra", "Doni"],
    "Umur" : [20, np.nan,22,100]
})

print("Data Awal")
print(df)

print("\nMissing")
print(df.isnull().sum)

df["Umur"] = df ["Umur"].fillna(df["Umur"].mean())

print("\nSetelah Cleaning")
print(df)

print("\nEkstraksi Data")
print(df[df["Umur"] >21])
```

## Kesimpulan

Pada pertemuan ini saya mempelajari cara mendeteksi missing, membersihkan data, dan melakukan ekstraksi data berdasarkan kondisi tertentu.
