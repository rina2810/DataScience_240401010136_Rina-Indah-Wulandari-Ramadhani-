## Topik

Pengantar Machine Learning : Regresi Linier










```python
import pandas as pd
from sklearn.linear_model import LinearRegression

df = pd.DataFrame ({
    "Jam_Belajar" : [1,2,3,4,5],
    "Nilai" : [60,65,75,85,95]

})

x = df[["Jam_Belajar"]]
y = df["Nilai"]

model = LinearRegression()
model.fit(x,y)

prediksi = model.predict([[6]])

print("Prediksi nilai untuk 6 jam belajar:")
print(prediksi[0])
```

## Kesimpulan

Pada pertemuan ini saya mempelajari konsep dasar machine learning menggunakan regresi linear untuk melakukan prediksi berdasarkan data yang tersedia.
